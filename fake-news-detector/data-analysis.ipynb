```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Check for duplicate rows in the DataFrame
duplicates = true_df.duplicated()

# Display the rows that are duplicates
true_df[duplicates]
```

```python
# Add target label
fake_df['target'] = 1
true_df['target'] = 0
```
