Simple example for computing item-item similarities for Collaborative filtering as adopted from https://towardsdatascience.com/user-to-user-vs-item-to-item-collaborative-filtering-explained-through-simple-examples-part-1-f133bec23a58

```python
import pandas as pd
```

```python
df = pd.DataFrame({'u1':[5,1,4],'u2':[1,5,1]},index=['m1','m2','m3'])
df
```

```python
# computing item-item similarity between the three movies
from sklearn.metrics.pairwise import cosine_similarity
df_cs = pd.DataFrame(data=cosine_similarity(df),
                     columns=['m1','m2','m3'],
                     index=['m1','m2','m3'])
df_cs
```

A common approach to normalizing how “difficult a reviewer” a user may be in comparison to other users is to normalize their ratings by subtracting the average value and dividing the ratings by the range.

```python
def normalize(row):
    row_norm = (row-row.mean())/(row.max()-row.min())
    return row_norm

df_norm = df.apply(normalize)
df_norm
```

```python
# Compute cosine similarity on normalizing
def normalize(row):
    row_norm = (row-row.mean())/(row.max()-row.min())
    return row_norm

df_norm = df.apply(normalize)

df_cs_norm = pd.DataFrame(data=cosine_similarity(df_norm),
                          columns=['m1','m2','m3'],
                          index=['m1','m2','m3'])

df_cs_norm
```
